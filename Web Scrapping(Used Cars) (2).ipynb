```python
import pandas as pd
from selenium import webdriver
import time
from bs4 import BeautifulSoup
```

```python
url = "https://www.cars24.com/buy-used-cars-bangalore"
```

```python
browser = webdriver.Chrome()
browser.get(url)
time.sleep(5)

while True:
    cars = browser.find_elements("css selector", "div.styles_outer__NTVth")
    if len(cars) == 0:
        break

    last_car = cars[-1]

    browser.execute_script("arguments[0].scrollIntoView({block: 'end'});",last_car)

    time.sleep(3)

    new_cars = browser.find_elements("css selector", "div.styles_outer__NTVth")

    if len(new_cars) == len(cars):
        break                        
html=browser.page_source
```

```python
soup= BeautifulSoup(html,"html.parser")
soup
```

```python
cars = soup.find_all('div',class_="styles_outer__NTVth")
len(cars)
```

```python
car=cars[0]
```

```python
print(car.prettify())
```

```python
name=car.find('span',class_="sc-bcXHqh bAcffq").get_text()
name
```

```python
Manufacturing_Year=name.split()[0]
Manufacturing_Year
```

```python
Company=name.split()[1]
Company
```

```python
ul=car.find('ul')
items=ul.find_all("p")
items
```

```python
KM_Driven=items[0].get_text()
KM_Driven
```

```python
Fuel=items[1].get_text()
Fuel
```

```python
Gear_Type = items[2].get_text()
Gear_Type
```

```python
div=car.find('div',class_='styles_flexItem__2z__J')
emi=div.find('span')
if emi:
    EMI=emi.get_text()
else:
    None
print(EMI)
```

```python
Price=car.find('p',class_="sc-bcXHqh hvRpEM").get_text()
Price
```

```python
Manufacturing_Year_list=[]
Company_list=[]
Car_Name_list=[]
KM_Driven_list=[]
Fuel_list=[]
Gear_Type_list=[]
EMI_list=[]
price_list=[]
```

```python
for i in cars :
    name=i.find('span').get_text()
    Manufacturing_Year=name.split()[0]
    Company=name.split()[1]
    Car_Name=" ".join(name.split()[1:])
    
    ul=i.find('ul')
    items=ul.find_all("p")
    
    KM_Driven=items[0].get_text()
    
    Fuel=items[1].get_text()
    
    Gear_Type=items[2].get_text()

    div=i.find('div',class_='styles_flexItem__2z__J')
    emi=div.find('span')
    if emi:
        EMI=emi.get_text()
    else:
        None
    
    price_wrap = i.find("div", class_="styles_priceWrap__VwWBV")
    new_price = None

    if price_wrap:
        prices = price_wrap.find_all("p")

        if len(prices) == 1:
            new_price = prices[0].get_text(strip=True)

        elif len(prices) >= 2:
            new_price = prices[1].get_text(strip=True)

    price=new_price
    
    Manufacturing_Year_list.append(Manufacturing_Year)
    Company_list.append(Company)
    Car_Name_list.append(Car_Name)
    KM_Driven_list.append(KM_Driven)
    Fuel_list.append(Fuel)
    Gear_Type_list.append(Gear_Type)
    EMI_list.append(EMI)
    price_list.append(price)
```

```python
df=pd.DataFrame({'Car Name':Car_Name_list,
                 'Company':Company_list,
                 'Year of Manufacture':Manufacturing_Year_list,
                 'Distance Driven':KM_Driven_list,
                 'Fuel Type':Fuel_list,
                 'Gear Type':Gear_Type_list,
                 'EMI':EMI_list,
                 'Price':price_list
                })
```

```python
df.head(5)
```

```python
df.to_csv('Banglore_Cars24.csv',index=False)
```
